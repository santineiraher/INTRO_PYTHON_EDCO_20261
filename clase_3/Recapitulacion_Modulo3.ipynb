## 3. Limpieza de datos

### Conversión de fechas
Al importar datos desde Excel, las fechas pueden llegar como números enteros (por ejemplo, `20190609` para 9 de junio de 2019) en lugar del tipo `datetime`. Para poder filtrar, ordenar y realizar operaciones con fechas, debemos convertirlas usando `pd.to_datetime()`.

**Parámetro `format`**: Indica el formato en que vienen las fechas:
- `%Y` — año con 4 dígitos (ej: 2019)
- `%m` — mes con 2 dígitos (ej: 06)
- `%d` — día con 2 dígitos (ej: 09)

En este caso, `'%Y%m%d'` indica que la fecha está como `AAAAMMDD`.

```python
productos_consumo_masivo['date'].head()
```

```python
productos_consumo_masivo['date'] = pd.to_datetime(productos_consumo_masivo['date'], format='%Y%m%d')
```

```python
productos_consumo_masivo['date'].head()
```

### Conversión de precios
La columna `prod_unit_price` viene como texto con comas como separador decimal. Debemos reemplazar la coma por punto y convertir a numérico con `pd.to_numeric()`.

```python
productos_consumo_masivo['prod_unit_price'].dtype
```

```python
productos_consumo_masivo['prod_unit_price'] = pd.to_numeric(
    productos_consumo_masivo['prod_unit_price'].astype(str).str.replace(',', '.'),
    errors='coerce'
)
```
